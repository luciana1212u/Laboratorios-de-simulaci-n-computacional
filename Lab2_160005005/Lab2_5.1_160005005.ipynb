## 5.1.1 Generador de cuadrados medios (MidSquare)

Tomando en cuenta el tema visto en clase en las diapositivas de Números Aleatorios implemente o use un
generador con el método de MidSquare o Middle-Square (o cuadrados medios) para generar una secuencia ui
= u1, u2, …, un de n =100 números pseudoaleatorios usando las siguientes semillas:


```python
def midSquare(xant, d):

    cuadrado = str(xant**2).zfill(2*d)
    ini = (2*d - d)//2
    xsig = int(cuadrado[ini:ini+d])
    usig = xsig / (10**d)
    return [xsig, usig]

def midSquareN(x0, N):

    d = len(str(x0))
    xant = x0
    I, X, U = [], [], []
    for t in range(1, N+1):
        [xi, ui] = midSquare(xant, d)
        I.append(t); X.append(xi); U.append(ui)
        xant = xi
    return [I, X, U]
```

```python
semillas_ms = [1931, 675248, 4567, 12122, 7179345]

for x0 in semillas_ms:
    d = len(str(x0))
    [I, X, U] = midSquareN(x0, N)
    nombre = "MidSquare X0=%d" % x0
    SECUENCIAS[nombre] = {
        'U': U, 'X': X,
        'estado0': x0,
        'siguiente': (lambda d: (lambda x: midSquare(x, d)[0]))(d)
    }
    print("="*70)
    print("%s   (d = %d dígitos)" % (nombre, d))
    print("Primeros 10 valores:")
    showValues(I[:10], X[:10], U[:10])
    print("u_i (primeros 15):", np.round(U[:15], 4))
```

## 5.1.2 Generador congruencial mixto (GLC)

Tomando en cuenta el tema visto en clase en las diapositivas de Números Aleatorios y el Capítulo 3 Números
Aleatorios de [Ross99], implemente o use el método de generador congruencial mixto o generador lineal
congruencial (GLC) de acuerdo con la siguiente ecuación:

$$X_{n+1} = (a X_n + c) \bmod m, \qquad u_i = \frac{X_i}{m}$$
Para generar una secuencia ui = u1, u2, …, un de n=100 números pseudoaleatorios


```python
def genran(a, c, m, xant):
    xsig = (a*xant + c) % m
    usig = xsig / m
    return [xsig, usig]

def genranN(a, c, m, x0, N):
    xant = x0
    I = []
    X = []
    U = []
    for t in range(1, N+1):
        [xi, ui] = genran(a, c, m, xant)
        I.append(t)
        X.append(xi)
        U.append(ui)
        xant = xi
    return [I, X, U]
```

```python
# (etiqueta, X0, a, c, m)
params_glc = [
    ('a', 127,      115,         0,  128),
    ('b', 43,       25,          5,  48),
    ('c', 13,       16807,       0,  2**31 - 1),
    ('d', 29022024, 25214903917, 11, 2**48),
    ('e', 3,        5,           7,  200),
]

for etq, x0, a, c, m in params_glc:
    [I, X, U] = genranN(a, c, m, x0, N)
    nombre = "GLC(%s) X0=%d,a=%d,c=%d,m=%d" % (etq, x0, a, c, m)
    SECUENCIAS[nombre] = {
        'U': U, 'X': X,
        'estado0': x0,
        'siguiente': (lambda a, c, m: (lambda x: (a*x + c) % m))(a, c, m)
    }
    print("="*70)
    print(nombre)
    print("Primeros 10 valores:")
    showValues(I[:10], X[:10], U[:10])
    print("u_i (primeros 15):", np.round(U[:15], 4))
```

## 5.1.3 Otros generadores de números aleatorios

Tomando en cuenta la lectura del artículo científico [Mancilla2000], implemente al menos dos (2) métodos de los
siguientes generadores de numeros pseudoaleatorios y genere una secuencia ui = u1, u2, …, un de n =100
números aleatorios con unos parámetros de ejemplo que Ud. escoja.


### a. Generador Shift-Register

$$X_{n+k} = \left(\sum_{i=0}^{k-1} a_i X_{n+i}\right) \bmod 2$$

donde los $X_n$ y los $a_i$ son 0 o 1, y el período máximo es $2^{k}-1$. Se usa la realimentación de dos términos $x_n = (x_{n-k} + x_{n-q}) \bmod 2$ con $k = 31$ y $q = 3$ (polinomio primitivo $x^{31}+x^{3}+1$), en la variante **multi-step**: se toman $L = 31$ bits sucesivos para formar cada entero.

### b. Generador Lagged-Fibonacci

$$X_n = (X_{n-j} \oplus X_{n-k}) \bmod M, \quad j < k, \; M = 2^{m}$$

con el operador $\oplus$ = suma, retardos $j = 5$, $k = 17$ y $M = 2^{31}$. La tabla inicial de $k$ valores se llena con un GLC.



```python
# a. Shift-Register
K_SR, Q_SR, L_SR = 31, 3, 31

def paso_shift_register(estado):

    bits = list(estado)
    for _ in range(L_SR):
        nuevo = bits[0] ^ bits[K_SR - Q_SR]
        bits.pop(0)
        bits.append(nuevo)
    return tuple(bits)

def genShiftRegisterN(estado0, N):
    estado = tuple(estado0)
    X, U = [], []
    for _ in range(N):
        estado = paso_shift_register(estado)
        val = 0
        for b in estado[-L_SR:]:
            val = (val << 1) | b
        X.append(val)
        U.append(val / (2**L_SR))
    return X, U


estado0_sr = tuple(int(b) for b in format(29022024, '031b'))
X_sr, U_sr = genShiftRegisterN(estado0_sr, N)

SECUENCIAS["Shift-Register k=31,q=3"] = {
    'U': U_sr, 'X': X_sr, 'estado0': estado0_sr, 'siguiente': paso_shift_register
}
print("Shift-Register -> u_i (primeros 15):", np.round(U_sr[:15], 4))
```

```python
# ---------- b. Lagged-Fibonacci ----------
J_LF, K_LF, M_LF = 5, 17, 2**31

def paso_lagged_fibonacci(estado):
    """estado = tupla con los últimos K_LF valores (el más antiguo primero)."""
    tabla = list(estado)
    nuevo = (tabla[K_LF - J_LF] + tabla[0]) % M_LF   # X_{n-j} + X_{n-k}
    tabla.pop(0)
    tabla.append(nuevo)
    return tuple(tabla)

def genLaggedFibonacciN(estado0, N):
    estado = tuple(estado0)
    X, U = [], []
    for _ in range(N):
        estado = paso_lagged_fibonacci(estado)
        X.append(estado[-1])
        U.append(estado[-1] / M_LF)
    return X, U


[_, X_ini, _] = genranN(1103515245, 12345, M_LF, 29022024, K_LF)
estado0_lf = tuple(X_ini)
X_lf, U_lf = genLaggedFibonacciN(estado0_lf, N)

SECUENCIAS["Lagged-Fibonacci j=5,k=17"] = {
    'U': U_lf, 'X': X_lf, 'estado0': estado0_lf, 'siguiente': paso_lagged_fibonacci
}
print("Lagged-Fibonacci -> u_i (primeros 15):", np.round(U_lf[:15], 4))
```

### (a) Tamaño de ciclo o período




```python
def detectar_ciclo(estado0, siguiente, max_iter=200000):

    vistos = {}
    estado = estado0
    for i in range(max_iter):
        if estado in vistos:
            mu = vistos[estado]
            lam = i - mu
            return mu, lam
        vistos[estado] = i
        estado = siguiente(estado)
    return None, None

# tope de iteraciones
TOPES = {"Shift-Register k=31,q=3": 20000}

periodos = {}
for nombre, s in SECUENCIAS.items():
    tope = TOPES.get(nombre, 200000)
    mu, lam = detectar_ciclo(s['estado0'], s['siguiente'], max_iter=tope)
    periodos[nombre] = (mu, lam, tope)
    if lam is None:
        print("%-34s  período > %d (no se detecta dentro del tope)" % (nombre, tope))
    else:
        deg = "sí (se degenera)" if mu > 0 else "no"
        print("%-34s  período λ = %-8d transitorio μ = %-6d ¿degenera?: %s" % (nombre, lam, mu, deg))
```

### (b) Prueba de uniformidad 1: contraste $\chi^2$

Se divide $(0,1)$ en $k$ clases de igual amplitud; con $f_i$ observaciones en la clase $i$ y $e_i = n/k$ esperadas bajo $H_0$ (uniformidad),

$$\chi^{2}=\sum_{i=1}^{k}\frac{(f_{i}-e_{i})^{2}}{e_{i}} \sim \chi^{2}_{k-1}$$

Se rechaza la hipótesis de uniformidad si $\chi^{2} > \chi^{2}_{k-1,\alpha}$. Con $n = 100$ se toma $k = 10$, de modo que $e_i = 10 > 5$ y la aproximación asintótica es válida.

```python
def test_chi2_uniformidad(U, k=10, alpha=ALPHA):
    n = len(U)
    f, _ = np.histogram(U, bins=k, range=(0.0, 1.0))
    e = n/k
    T = np.sum((f - e)**2 / e)
    Tval = stats.chi2.ppf(1 - alpha, k - 1)
    pvalue = 1 - stats.chi2.cdf(T, k - 1)
    return T, Tval, pvalue, (T > Tval)

nombre_demo = list(SECUENCIAS.keys())[0]
T, Tval, pv, rechaza = test_chi2_uniformidad(SECUENCIAS[nombre_demo]['U'])
print("Secuencia:", nombre_demo)
print("Chi^2 = %.4f   Chi^2_{k-1,alpha} = %.4f   p-value = %.4f" % (T, Tval, pv))
print("Se rechaza la hipotesis de uniformidad." if rechaza else "Se acepta la hipotesis de uniformidad.")
```

### (c) Prueba de aleatoriedad: test de rachas

Dada la muestra $u_1,\dots,u_n$ se construye una sucesión de 0 y 1: en la posición $i$ se coloca 0 si $u_{i+1} < u_i$ y 1 si $u_{i+1} > u_i$. Cada grupo consecutivo de 0's o de 1's es una **racha**. Para $n$ grande el número de rachas $R$ sigue

$$R \sim \mathcal{N}\left(\frac{2n-1}{3},\; \sigma^{2}=\frac{16n-29}{90}\right), \qquad Z = \frac{R-\frac{2n-1}{3}}{\sqrt{\frac{16n-29}{90}}}\sim\mathcal{N}(0,1)$$

Se rechaza la hipótesis de aleatoriedad si $|Z| > Z_{\alpha/2}$.

>

```python
def test_rachas(U, alpha=ALPHA):
    n = len(U)
    A = np.zeros(n-1)
    for i in range(1, n):
        if U[i-1] > U[i]:
            A[i-1] = 0
        else:
            A[i-1] = 1

    A2 = np.array((A == 1))
    sw = (A2[:-1] ^ A2[1:])
    isw = np.arange(len(sw))[sw]
    isw = list(isw); isw.insert(0, -1)
    isw = np.array(isw)
    L = isw[1:len(isw)] - isw[0:len(isw)-1]

    R = len(L) + 1
    Z = (R - ((2*n - 1)/3)) / (np.sqrt((16*n - 29)/90))
    Zt = stats.norm.ppf((1 + (1 - alpha))/2)
    return R, Z, Zt, L, (abs(Z) > Zt)

R, Z, Zt, L, rechaza = test_rachas(SECUENCIAS[nombre_demo]['U'])
print("Secuencia:", nombre_demo)
print("Longitudes de las primeras rachas:", L[:20])
print("R = %d   Z = %.4f   Z_{alpha/2} = %.4f" % (R, Z, Zt))
print("Se rechaza la hipotesis de aleatoriedad." if rechaza else "Se acepta la hipotesis de aleatoriedad.")
```

#### Criterios usados para clasificarlos

Siguiendo a [Mancilla2000], un buen generador debe cumplir: **período máximo**, **uniformidad**, **independencia / secuencias no correlacionadas**, **eficiencia** de cómputo y **reproducibilidad**. En este laboratorio el criterio que más discrimina no son las pruebas estadísticas sino el **tamaño de ciclo**: todos los generadores rechazados o dudosos tienen períodos ridículamente cortos o se degeneran, mientras que los generadores con módulo grande y parámetros que cumplen las condiciones de período máximo pasan todas las pruebas.
